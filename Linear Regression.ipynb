## Explain the main assumptions of Linear Regression in detail.

1. Linerity : The relationship between each independent variable and the dependent variable should be linear, given the other predictors in the model.
Example: As experience increases by one year, predicted salary increases by approximately ₹1 lakh, assuming other factors remain constant.
2. No Multi colinarity: Independent variables should not have perfect or excessively high correlations with one another.
Example: Age and experience may be highly correlated because older employees often have more experience. This can make it difficult to distinguish their individual effects on salary.
3. Homoscedacity: The variance (spread) of the errors should remain approximately constant across different predicted values.
Example: Salary predictions should have a similar degree of error spread for both lower-paid and higher-paid employees.
4. No Heteroscedacity: Heteroscedasticity occurs when the variance of errors changes across predicted values. Ordinary least squares regression generally assumes homoscedasticity for its usual standard errors and statistical tests. Errors forms a funnel shape.
5. Errors should be normally distributed.the errors are assumed to be normally distributed around zero, conditional on the predictors.

## B.	What is the difference between R-squared and Adjusted R-squared?

R² is calculated using the sum of squared errors between actual and predicted values, compared with the total variation in actual values around their mean.R² tells us how much of the variation in the target variable is explained by the model relative to a baseline that predicts the mean.

Adjusted R² is R² adjusted for the number of independent variables and the number of observations. It penalizes adding predictors that do not improve the model sufficiently.

## C. Different Types of Regularization Techniques in Regression

There are three main types of regularization techniques in regression:

1. Ridge Regression (L2 Regularization)
2. Lasso Regression (L1 Regularization)
3. Elastic Net Regression (combination of L1 and L2)

**Why is regularization required?**

For example, a model predicts salary based on experience, age, education, etc. The model may assign large coefficients to features to fit the training data very closely. This can cause overfitting, meaning the model performs well on training data but poorly on testing data.

Regularization adds a penalty for large coefficients to the cost function. This helps reduce overfitting and improves the model's ability to generalize to new data.

### 1. Ridge Regression (L2 Regularization)

Ridge Regression calculates the penalty by squaring the coefficients.

**Cost function:**

Cost = MSE + λ × Sum of squared coefficients

Example: If the coefficients are 3 and 4:

Penalty = 3² + 4² = 9 + 16 = 25

If MSE = 10 and λ = 1:

Cost = 10 + (1 × 25) = 35

Ridge reduces the coefficient values towards zero but generally does not make them exactly zero. Therefore, it keeps all features in the model.

### 2. Lasso Regression (L1 Regularization)

Lasso Regression calculates the penalty by taking the absolute values of the coefficients.

**Cost function:**

Cost = MSE + λ × Sum of absolute coefficients

Example: If the coefficients are 3 and 4:

Penalty = |3| + |4| = 3 + 4 = 7

If MSE = 10 and λ = 1:

Cost = 10 + (1 × 7) = 17

Lasso reduces coefficient values and can make some coefficients exactly zero. Therefore, it can remove less important features from the model.

### 3. Elastic Net Regression

Elastic Net combines Ridge (L2) and Lasso (L1) regularization.

**Cost function:**

Cost = MSE + λ₁ × Sum of absolute coefficients + λ₂ × Sum of squared coefficients

Example: If the coefficients are 3 and 4, MSE = 10, and λ₁ = λ₂ = 1:

L1 penalty = |3| + |4| = 7

L2 penalty = 3² + 4² = 25

Total cost = 10 + 7 + 25 = 42

Elastic Net both reduces coefficient values and can make some coefficients exactly zero. It combines the benefits of Ridge and Lasso.

### Conclusion

Ridge reduces coefficient values, Lasso can remove features by making coefficients zero, and Elastic Net combines both techniques. All three techniques help reduce overfitting.

## D. How Logistic Regression Works for Multiclass Classification

**Multiclass classification** is used when we want to predict one category out of more than two categories.

For example, predicting a fruit based on its weight, color, and size:
- Apple
- Orange
- Banana

Logistic Regression can handle multiclass classification using two common approaches: **One-vs-Rest (OvR)** and **Softmax Regression**.

### 1. How it works

**Step 1: Take input features**

The model takes input features such as weight, color, and size.

**Step 2: Calculate scores**

The model calculates a score for each class using the input features and learned coefficients.

Score = (Coefficient × Feature) + Bias

**Step 3: Convert scores into probabilities**

The model converts the scores into probabilities between 0 and 1.

In Softmax Regression, the probabilities of all classes add up to 1.

**Step 4: Select the class**

The model predicts the class with the highest probability.

### 2. Softmax Regression

Softmax Regression is a common approach for multiclass classification. It converts the scores of all classes into probabilities.


**Example:**

Suppose the model predicts the following probabilities for a fruit:

| Fruit | Predicted probability |
|---|---:|
| Apple | 0.70 |
| Orange | 0.20 |
| Banana | 0.10 |
| Total | 1.00 |

The model predicts **Apple** because it has the highest probability, 70%.

### 3. One-vs-Rest (OvR)

In One-vs-Rest, the model creates one binary classifier for each class.

For example, for three fruit classes:

1. Apple vs all other fruits.
2. Orange vs all other fruits.
3. Banana vs all other fruits.

Each classifier calculates the probability that the input belongs to its particular class. The model selects the class with the highest score or probability.

### 4. How the model learns

During training, the model compares its predicted probabilities with the actual classes using a loss function, commonly cross-entropy loss.

The model adjusts its coefficients and bias to reduce the loss. This process continues until the model learns parameters that fit the training data.

### 5. Advantages

- Can classify data into more than two categories.
- Produces probabilities for each class.
- Easy to understand and interpret.
- Works well when classes can be separated reasonably well using linear decision boundaries.

### Conclusion

Logistic Regression performs multiclass classification by calculating a score for each class, converting the scores into probabilities using Softmax or One-vs-Rest classifiers, and predicting the class with the highest probability. During training, it learns coefficients that minimize the classification loss.

## E.	Explain the performance metrics of logistic regression.

1. Accuracy
Accuracy tells us how many predictions made by the model are correct out of all predictions.
Formula:
Accuracy = (TP + TN) /(TP+TN+FP+FN)

2. Precision
Precision tells us out of all the positive predictions made by the model, how many are actually positive.
Formula:
 
Precesion = (TP)/(TP+FP)

Example: The model predicts that 10 emails are spam, but only 8 are actually spam. Precision = 8/10 = 80%.

3. Recall
Recall tells us out of all the actual positive cases, how many the model correctly identifies.
Formula: TP/(TP+FN)

Example: There are 10 actual spam emails, and the model identifies 8 of them. Recall = 8/10 = 80%.

4. F1 Score
F1 score is the harmonic mean of precision and recall. It balances both metrics and is useful when we want to consider both false positives and false negatives.
Formula: 2*(Precesion * Recall)/Precesion + Recall


Example: If precision is 80% and recall is 60%:
F1 SCore = 2*(80*60/80+60 )= 2*(4800/140) = 34.28*2 = 68.57

Remember these terms
Term	Meaning
TP — True Positive	Correctly predicted positive
TN — True Negative	Correctly predicted negative
FP — False Positive	Predicted positive but actually negative
FN — False Negative	Predicted negative but actually positive
